```python
# Step 1: Import pandas library (for working with tables)
import pandas as pd

# Step 2: Load the dataset from the data folder
df = pd.read_csv('../data/netflix_titles.csv')  # .. means go up one folder

# Step 3: See the first 5 rows of data
df.head()
```

```python
# Show rows where any value is missing
df[df.isna().any(axis=1)].head()
```

```python
df = df.dropna(subset=['title', 'type'])  # drop rows if title or type is missing
```

```python
# Convert date_added column to datetime format
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
# Remove spaces at the start and end of the string
df['date_added'] = df['date_added'].str.strip()
```

```python
# Convert to datetime, ignore rows that cannot be converted
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['date_added'].head(10)
df['date_added'].isna().sum()  # count how many rows became NaT
```

```python
# Count how many Movies and TV Shows
df['type'].value_counts()
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar', color=['skyblue', 'orange'])
plt.title('Movies vs TV Shows on Netflix')
plt.ylabel('Count')
plt.show()
```

```python
df['country'].value_counts().head(10)  # Top 10 countries
```

```python
df['country'].value_counts().head(10).plot(kind='barh', color='green')
plt.title('Top 10 Countries with Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
# Split the 'listed_in' column (genres) by comma
genres = df['listed_in'].str.split(', ').explode()  # explode makes one row per genre

# Count top 10 genres
genres.value_counts().head(10)
```

```python
genres.value_counts().head(10).plot(kind='bar', color='purple')
plt.title('Top 10 Genres on Netflix')
plt.ylabel('Count')
plt.show()
```

```python
# Create a new column 'year_added' from 'date_added'
df['year_added'] = df['date_added'].dt.year

# Count titles per year
df['year_added'].value_counts().sort_index()  # sort by year

```

```python
df['year_added'].value_counts().sort_index().plot(kind='line', marker='o', color='red')
plt.title('Titles Added per Year on Netflix')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# Count each rating
df['rating'].value_counts()
```

```python
df['rating'].value_counts().plot(kind='bar', color='teal')
plt.title('Netflix Titles by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# Split actors and explode to make one row per actor
actors = df['cast'].dropna().str.split(', ').explode()

# Count top 10 actors
top_actors = actors.value_counts().head(10)
top_actors
```

```python
top_actors.plot(kind='barh', color='orange')
plt.title('Top 10 Actors on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Actor')
plt.show()
```

```python
rating_type = df.groupby(['rating', 'type']).size().unstack()
rating_type.plot(kind='bar', stacked=True, figsize=(10,5))
plt.title('Movies vs TV Shows by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# Save cleaned CSV file
df.to_csv('../data/netflix_clean.csv', index=False)
```
